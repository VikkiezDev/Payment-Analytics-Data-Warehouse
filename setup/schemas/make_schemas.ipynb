```sql
-- Use the paymentdw catalog (created 'paymentdw')
CREATE SCHEMA IF NOT EXISTS paymentdw.bronze
COMMENT 'Raw data landed as-is - never modified';

CREATE SCHEMA IF NOT EXISTS paymentdw.silver
COMMENT 'Cleaned, typed, deduplicated data';

CREATE SCHEMA IF NOT EXISTS paymentdw.gold
COMMENT 'Star schema - fact and dimension tables for BI';

CREATE SCHEMA IF NOT EXISTS paymentdw.logs
COMMENT 'Pipeline run logs and data quality failures';

-- Verify
SHOW SCHEMAS IN paymentdw;
```
