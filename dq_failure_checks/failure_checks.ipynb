```sql
-- ═══════════════════════════════════════════════════════════
-- DATA QUALITY CHECKS
-- Runs assertions against Silver tables
-- Logs every check (PASSED or FAILED) to dq_failures
-- ═══════════════════════════════════════════════════════════

-- ── CHECK 1: Null primary keys in orders ─────────────────
INSERT INTO paymentdw.logs.dq_failures
    (layer, table_name, check_name, check_type, failed_row_count, threshold, status, run_timestamp)
SELECT
    'silver',
    'orders',
    'null_check_order_id',
    'NULL_CHECK',
    SUM(CASE WHEN order_id IS NULL THEN 1 ELSE 0 END),
    '0 nulls allowed',
    CASE WHEN SUM(CASE WHEN order_id IS NULL THEN 1 ELSE 0 END) = 0
         THEN 'PASSED' ELSE 'FAILED' END,
    current_timestamp()
FROM paymentdw.silver.orders;
```

```sql
-- ── CHECK 14: Orphaned order_items (no parent order) ─────
INSERT INTO paymentdw.logs.dq_failures
    (layer, table_name, check_name, check_type, failed_row_count, threshold, status, run_timestamp)
SELECT
    'silver',
    'order_items',
    'referential_integrity_order_id',
    'REFERENTIAL_INTEGRITY',
    COUNT(*),
    '0 orphaned records allowed',
    CASE WHEN COUNT(*) = 0
         THEN 'PASSED' ELSE 'FAILED' END,
    current_timestamp()
FROM paymentdw.silver.order_items i
LEFT JOIN paymentdw.silver.orders o
    ON i.order_id = o.order_id
WHERE o.order_id IS NULL;
```

```sql
-- ── CHECK 15: Future purchase dates in orders ─────────────
INSERT INTO paymentdw.logs.dq_failures
    (layer, table_name, check_name, check_type, failed_row_count, threshold, status, run_timestamp)
SELECT
    'silver',
    'orders',
    'range_check_no_future_dates',
    'RANGE_CHECK',
    SUM(CASE WHEN purchase_timestamp > current_timestamp()
             THEN 1 ELSE 0 END),
    '0 future purchase dates allowed',
    CASE WHEN SUM(CASE WHEN purchase_timestamp > current_timestamp()
                       THEN 1 ELSE 0 END) = 0
         THEN 'PASSED' ELSE 'FAILED' END,
    current_timestamp()
FROM paymentdw.silver.orders;
```

```sql
-- ── VIEW ALL DQ RESULTS ───────────────────────────────────
SELECT
    layer,
    table_name,
    check_name,
    check_type,
    failed_row_count,
    threshold,
    status,
    run_timestamp
FROM paymentdw.logs.dq_failures
ORDER BY status DESC, table_name, check_name;
```

```sql
-- What do negative amount transactions look like?
SELECT
    error_type,
    use_chip                            AS payment_channel,
    COUNT(*)                            AS txn_count,
    ROUND(SUM(amount), 2)               AS total_amount,
    ROUND(AVG(amount), 2)               AS avg_amount,
    MIN(amount)                         AS min_amount,
    MAX(amount)                         AS max_amount
FROM paymentdw.silver.transactions
WHERE amount < 0
GROUP BY error_type, use_chip
ORDER BY txn_count DESC;
```
