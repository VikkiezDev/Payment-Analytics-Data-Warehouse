## Data profiling Financial Transactions Dataset

```sql
select * from cards_data limit 5;
```

```sql
select
    count(*) as total_rows,
    sum(case when id is null then 1 else 0 end) as null_id,
    sum(case when client_id is null then 1 else 0 end) as null_client_id,
    count(distinct card_type) as unique_card_types,
    count(distinct card_number) as unique_card_numbers,
    sum(case when try_cast(regexp_replace(credit_limit, '[$,]', '') as decimal(12,2)) = 0 then 1 else 0 end) as credit_limit_invalid
from 
    cards_data;
```

```sql
select * from users_data limit 5;
```

```sql
select
    count(*) as total_rows,
    count(distinct id) as unique_users,
    sum(case when current_age is null or current_age<=0 then 1 else 0 end) as invalid_age,
    count(distinct gender) as unique_gender,
    SUM(case when latitude is null then 1 else 0 end) as null_lat,
    SUM(CASE WHEN longitude is null then 1 else 0 end) as null_lon
from users_data;
```

```sql
select * from transactions_data limit 5;
```

```sql
select
    count(*) as total_rows,
    count(distinct id) as unique_txns,
    sum(case when amount is null then 1 else 0 end) as null_amounts,
    count(distinct use_chip) as distinct_channels,
    sum(case when merchant_city is null then 1 else 0 end) as null_city,
    sum(case when zip is null then 1 else 0 end) as null_zip,
    count(distinct errors) as distinct_error_types,
    sum(case when errors is null then 1 else 0 end) as null_errors
from transactions_data;
```

## Data Profiling: Brazilian E-Commerce Public Dataset by Olist

```sql
select * from olist_orders limit 5;
```

```sql
select
    count(*) as total_rows,
    count(distinct order_id) as unique_orders,
    sum(case when order_id is null then 1 else 0 end) as null_order_ids,
    sum(case when customer_id is null then 1 else 0 end) as null_customer_ids,
    sum(case when order_status is null then 1 else 0 end) as null_status,
    count(DISTINCT order_status) as distinct_statuses
from olist_orders;
```

```sql
select * from olist_order_items limit 5
```

```sql
select
    count(*) as total_rows,
    count(distinct order_id) as unique_orders,
    sum(case when price is null then 1 else 0 end) as null_prices,
    sum(case when price <= 0 then 1 else 0 end) as zero_or_neg_prices,
    min(price) as min_price,
    max(price) as max_price,
    avg(price) as avg_price
from olist_order_items;
```

```sql
select * from olist_order_payments limit 5;
```

```sql
select
    count(*) as total_rows,
    count(distinct order_id) as unique_orders,
    count(distinct payment_type) as distinct_payment_types,
    sum(case when payment_value is null then 1 else 0 end) as null_values,
    sum(case when payment_value <= 0 then 1 else 0 end) as zero_or_neg_values,
    min(payment_value) as min_value,
    max(payment_value) as max_value
from olist_order_payments;
```

```sql
select * from olist_customers limit 5;
```

```sql
SELECT
    count(*) as total_rows,
    count(distinct customer_id) as unique_customers,
    count(distinct customer_unique_id) as unique_real_customers,
    count(distinct customer_state) as distinct_states
FROM paymentdw.bronze.olist_customers;
```

```sql
select * from olist_sellers limit 5;
```

```sql
select
    count(*) as total_rows,
    count(distinct seller_id) as unique_sellers,
    count(distinct seller_state) as distinct_states,
    sum(CASE WHEN seller_zip_code_prefix is null then 1 else 0 end) as null_zips
from olist_sellers;
```

```sql
select * from olist_products limit 5;
```

```sql
select
    COUNT(*) as total_rows,
    COUNT(distinct product_id) as unique_products,
    SUM(case when product_category_name is null then 1 else 0 end) as null_categories,
    SUM(case when product_weight_g is null then 1 else 0 end) as null_weights,
    SUM(case when product_length_cm is null then 1 else 0 end) as null_dimensions
from olist_products;
```

```sql
select * from olist_reviews limit 5;
```

```sql
select
    count(*) as total_rows,
    count(distinct review_id) as unique_reviews,
    count(distinct order_id) as unique_orders,
    sum(case when review_score is null then 1 else 0 end) as null_scores,
    count(distinct review_score) as unique_scores,
    min(review_score) as min_score,
    max(review_score) as max_score,
    sum(case when review_comment_message is null then 1 else 0 end) as null_comments
from olist_reviews;
```

```sql
select
    count(*) as total_rows,
    count(distinct geolocation_zip_code_prefix) as unique_zip_codes,
    sum(case when geolocation_lat is null then 1 else 0 end) as null_lat,
    sum(case when geolocation_lng is null then 1 else 0 end) as null_lng,
    sum(case when geolocation_state is null then 1 else 0 end) as null_state,
    min(geolocation_lat) as min_lat,
    max(geolocation_lat) as max_lat,
    min(geolocation_lng) as min_lng,
    max(geolocation_lng) as max_lng
from olist_geolocation;
```

```sql
select * from product_category_translation limit 5;
```

```sql
select
    count(*) as total_rows,
    count(distinct product_category_name) as unique_portuguese_names,
    count(distinct product_category_name_english) as unique_english_names,
    sum(case when product_category_name is null then 1 else 0 end) as null_portuguese,
    sum(case when product_category_name_english is null then 1 else 0 end) as null_english
from product_category_translation;
```
