```sql
-- Pipeline logging table
create or replace table paymentdw.logs.pipeline_logs(
    log_id bigint generated always as identity,
    layer string,
    table_name string,
    source_table string,
    rows_read bigint,
    rows_written bigint,
    status string,
    error_message string,
    run_timestamp timestamp,
    duration_seconds int
);
```

```sql
create or replace table paymentdw.logs.dq_failures(
    dq_id bigint generated always as identity,
    layer string,
    table_name string,
    check_name string,
    check_type string,
    failed_row_count bigint,
    threshold string,
    status string,
    run_timestamp timestamp
);
```
